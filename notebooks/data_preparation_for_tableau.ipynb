```python
df = df.applymap(lambda x:x.strip('"') if isinstance(x,str)else x) 
```

```python
df.info()
```

```python
indicator_categories = {
    # 1. Gender Rights & Legal Equality
    **{name: "Gender Rights & Legal Equality" for name in df['indicator_name'] if name.startswith("A woman")},
    "Criminal penalties or civil remedies exist for sexual harassment in employment (1=yes; 0=no)": "Gender Rights & Legal Equality",
    "Women Business and the Law Index Score (scale 1-100)": "Gender Rights & Legal Equality",

    # 2. Political Participation
    "Proportion of seats held by women in national parliaments (%)": "Political Participation",
    "Proportion of women in ministerial level positions (%)": "Political Participation",

    # 3. Fertility & Demographics
    "Adolescent fertility rate (births per 1,000 women ages 15-19)": "Fertility & Demographics",
    "Birth rate, crude (per 1,000 people)": "Fertility & Demographics",
    "Life expectancy at birth, female (years)": "Fertility & Demographics",
    "Life expectancy at birth, male (years)": "Fertility & Demographics",

    # 4. Maternal & Reproductive Health
    "Maternal mortality ratio (modeled estimate, per 100,000 live births)": "Maternal & Reproductive Health",
    "Prevalence of HIV, female (% ages 15-24)": "Maternal & Reproductive Health",
    "Prevalence of HIV, male (% ages 15-24)": "Maternal & Reproductive Health",

    # 5. Education & Literacy
    "Educational attainment, at least completed upper secondary, population 25+, female (%) (cumulative)": "Education & Literacy",
    "Educational attainment, at least completed upper secondary, population 25+, male (%) (cumulative)": "Education & Literacy",
    "Literacy rate, adult female (% of females ages 15 and above)": "Education & Literacy",
    "Literacy rate, adult male (% of males ages 15 and above)": "Education & Literacy",
    "Literacy rate, adult total (% of people ages 15 and above)": "Education & Literacy",
    "Literacy rate, youth female (% of females ages 15-24)": "Education & Literacy",
    "Literacy rate, youth male (% of males ages 15-24)": "Education & Literacy",
    "School enrollment, secondary (gross), gender parity index (GPI)": "Education & Literacy",

    # 6. Labor & Employment
    "Employment in services, female (% of female employment) (modeled ILO estimate)": "Labor & Employment",
    "Employment in services, male (% of male employment) (modeled ILO estimate)": "Labor & Employment",
    "Labor force participation rate, female (% of female population ages 15+) (modeled ILO estimate)": "Labor & Employment",
    "Labor force participation rate, male (% of male population ages 15+) (modeled ILO estimate)": "Labor & Employment",
    "Unemployment, female (% of female labor force) (modeled ILO estimate)": "Labor & Employment",
    "Unemployment, male (% of male labor force) (modeled ILO estimate)": "Labor & Employment",

    # 7. Economic Context
    "GDP per capita (current US$)": "Economic Context",
    "Inflation, consumer prices (annual %)": "Economic Context",
}
```

```python
df['category'] = df['indicator_name'].map(indicator_categories)
df['category'] = df['category'].fillna('Other')
```

```python
df['category'].value_counts()
```

```python
df.to_csv('mena_cleaned_data1.csv',index=False)
```
